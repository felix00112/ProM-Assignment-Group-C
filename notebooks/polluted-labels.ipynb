## Polluted Labels
Problem: 
- Labels (i.e., attribute values) are structurally the same, yet are distinct from each other due to differences in the exact values
- Effect: Discovered process models will overfit the event-log, and be too little generic
  
Manifestation & Detection:
- Attribute values being composed of a mixture of immutable boiler-plate text (here: Notification of Loss, Incident No.)
and mutable text (here: XXXX, xxxxxx)
- May be detected if the different attribute values are higher than expected

```python
# add mutable text (EventID) to column concept:name
df["concept:name"] = (
    df["concept:name"].astype(str)
    + " - "
    + df["EventID"].astype(str)
)

df.head()
```
