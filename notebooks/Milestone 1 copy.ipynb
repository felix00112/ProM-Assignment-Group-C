```python
def pollute_labels(df, rate = 0.001):
    df = df.copy()
    n = int(len(df) * rate)
    indices = np.random.choice(df.index, n, replace=False)

    for index in indices:
        activity = df.loc[index, "concept:name"]
        case_id = df.loc[index, "case:concept:name"]
        df.loc[index, "concept:name"] = f"{activity} - Incident No. {case_id}"

    return df
```

```python
df_polluted = df.copy()
df_polluted = pollute_labels(df_polluted)
df_polluted.head(100)
```

```python
def distort_activity_label(label):
    replacements = {
        "application": "appl.",
        "Application": "App.",
        "offers": "ofrs.",
        "Offer": "Off.",
        "incomplete": "incompl.",
        "Incomplete": "Incompl.",
        "files": "docs.",
        "leads": "lds.",
        "Handle": "Hndl.",
        "Validate": "Valid.",
        "Validating": "Valid.",
        "Complete": "Comp.",
        "Created": "Crtd.",
        "Create": "Crt.",
        "Accepted": "Acc.",
        "Cancelled": "Canc.",
        "Submitted": "Subm.",
        "Pending": "Pend.",
        "Denied": "Den.",
        "Refused": "Ref.",
        "Returned": "Ret.",
        "potential": "pot.",
        "fraud": "frd.",
        "Shortened": "Short.",
        "completion": "compl.",
        "Personal": "Pers.",
        "Loan": "Ln.",
        "collection": "coll.",
        "online": "onl.",
        "mail": "email"
    }

    distorted = str(label)

    for original, replacement in replacements.items():
        if original in distorted:
            return distorted.replace(original, replacement, 1)

    return distorted + "."
```

```python
labels = df["concept:name"].dropna().unique()

test_df = pd.DataFrame({
    "original_label": labels,
    "distorted_label": [distort_activity_label(label) for label in labels]
})

test_df
```

```python
def pollute_distorted_labels(
    df,
    activity_column="concept:name",
    rate=0.05,
    pollution_column="pollution_type"
):
    df = df.copy()

    if pollution_column not in df.columns:
        df[pollution_column] = None

    n = int(len(df) * rate)
    indices = np.random.choice(df.index, n, replace=False)

    for index in indices:
        original_label = df.loc[index, activity_column]

        if pd.notna(original_label):
            distorted_label = distort_activity_label(original_label)

            df.loc[index, activity_column] = distorted_label
            df.loc[index, pollution_column] = "distorted_label"

    return df
```

```python
df_polluted = pollute_distorted_labels(
    df_polluted,
    activity_column="concept:name",
    rate=0.05
)
```

```python
def clean_polluted_labels(
    df,
    activity_column="concept:name",
    cleaning_column="label_cleaning"
):
    df = df.copy()

    if cleaning_column not in df.columns:
        df[cleaning_column] = None

    original_labels = df[activity_column].copy()

    # alles als string behandeln
    df[activity_column] = df[activity_column].astype("string")

    patterns = [
        r"\._\d+$",                              # ._1691306052
        r"_\d+$",                                # _1691306052
        r"\s*-\s*Incident No\.?\s*[A-Za-z0-9_\-]+",
        r"\s*-\s*Incident\s*[A-Za-z0-9_\-]+",
        r"\s*-\s*Case\s*[A-Za-z0-9_\-]+",
        r"\s*-\s*Application\s*[A-Za-z0-9_\-]+",
        r"\s*-\s*User\s*[A-Za-z0-9_\-]+",
        r"\s*-\s*Resource\s*[A-Za-z0-9_\-]+",
        r"\s*#\s*[A-Za-z0-9_\-]+",
        r"\s*\(\s*id\s*[:=]?\s*[A-Za-z0-9_\-]+\s*\)",
        r"\s*\(\s*case\s*[:=]?\s*[A-Za-z0-9_\-]+\s*\)",
    ]

    for pattern in patterns:
        df[activity_column] = df[activity_column].str.replace(
            pattern,
            "",
            regex=True
        )

    # Whitespace normalisieren
    df[activity_column] = (
        df[activity_column]
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )

    changed_mask = original_labels.astype("string") != df[activity_column]

    df.loc[changed_mask, cleaning_column] = "cleaned_polluted_label"

    return df
```

```python
df_cleaned = df_polluted.copy()

df_cleaned = clean_polluted_labels(
    df_cleaned,
    activity_column="concept:name"
)
```

```python
def clean_distorted_labels(
    df,
    activity_column="concept:name",
    cleaning_column="label_cleaning"
):
    df = df.copy()

    if cleaning_column not in df.columns:
        df[cleaning_column] = None

    original_labels = df[activity_column].copy()

    label_mapping = {
        "W_Validate appl.": "W_Validate application",
        "W_Call after ofrs.": "W_Call after offers",
        "W_Call incompl. docs.": "W_Call incomplete files",
        "W_Complete appl.": "W_Complete application",
        "W_Handle lds.": "W_Handle leads",

        "O_Create Off.": "O_Create Offer",
        "O_Crtd.": "O_Created",
        "O_Sent (email and onl.)": "O_Sent (mail and online)",
        "O_Sent (onl. only)": "O_Sent (online only)",
        "O_Ret.": "O_Returned",
        "O_Canc.": "O_Cancelled",
        "O_Acc.": "O_Accepted",
        "O_Ref.": "O_Refused",
        "O_Sent (mail and onl.)" : "O_Sent (mail and online)",

        "A_Valid.": "A_Validating",
        "A_Create App.": "A_Create Application",
        "A_Conc.": "A_Concept",
        "A_Acc.": "A_Accepted",
        "A_Comp.": "A_Complete",
        "A_Incompl.": "A_Incomplete",
        "A_Subm.": "A_Submitted",
        "A_Pend.": "A_Pending",
        "A_Canc.": "A_Cancelled",
        "A_Den.": "A_Denied",
        "A_Concept." : "A_Concept",

        "W_Assess pot. frd.": "W_Assess potential fraud",
        "W_Short. compl.": "W_Shortened completion",
        "W_Pers. Ln. coll.": "W_Personal Loan collection",
        "W_Call incompl. files" : "W_Call incomplete files",
        "W_Assess pot. fraud" : "W_Assess potential fraud",
        "W_Short. completion" : "W_Shortened completion",
        "W_Pers. Loan collection" : "W_Personal Loan collection",
    }

    df[activity_column] = df[activity_column].replace(label_mapping)

    changed_mask = original_labels.astype("string") != df[activity_column].astype("string")

    # nur markieren, wenn noch kein anderer label-cleaning Eintrag existiert
    df.loc[
        changed_mask & df[cleaning_column].isna(),
        cleaning_column
    ] = "cleaned_distorted_label"

    df.loc[
        changed_mask & df[cleaning_column].notna(),
        cleaning_column
    ] = df.loc[
        changed_mask & df[cleaning_column].notna(),
        cleaning_column
    ].astype(str) + "+cleaned_distorted_label"

    return df
```

```python
df_cleaned = clean_distorted_labels(
    df_cleaned,
    activity_column="concept:name"
)
```
