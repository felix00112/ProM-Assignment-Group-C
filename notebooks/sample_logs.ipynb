```python
import pandas as pd
import pm4py
from pm4py.objects.log.util import sampling

# parse logs
log_clean = pm4py.read_xes("../data/BPI Challenge 2017.xes.gz", return_legacy_log_object=True)
log_noised = pm4py.read_xes("../data/noised.xes.gz", return_legacy_log_object=True)
log_recovered = pm4py.read_xes("../data/recovered.xes.gz", return_legacy_log_object=True)
```

```python
no_traces = 5000

log_clean_sample = sampling.sample_log(log_clean, no_traces)
print("Number of traces (sampled clean log):", len(log_clean_sample))
```

```python
log_noised_sample = sampling.sample_log(log_noised, no_traces)
print("Number of traces (sampled noised log):", len(log_noised_sample))
```

```python
log_recovered_sample = sampling.sample_log(log_recovered, no_traces)
print("Number of traces (sampled recovered log):", len(log_recovered_sample))
```

```python
pm4py.write_xes(log_clean_sample, "../data/samples/clean_sample.xes.gz")
pm4py.write_xes(log_noised_sample, "../data/samples/noised_sample.xes.gz")
pm4py.write_xes(log_recovered_sample, "../data/samples/recovered_sample.xes.gz")
```
