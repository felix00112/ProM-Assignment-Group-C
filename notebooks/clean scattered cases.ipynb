```python
import pandas as pd
import pm4py
from pathlib import Path

# read noised log to dataframe
noised_log = pm4py.read_xes("../data/noised.xes")
noised_df = pm4py.convert_to_dataframe(noised_log)

# read CSV containing the scattered rows
validation_df = pd.read_csv("../data/scattered_events_validation.csv")
```

```python
# reindex
validation_df = validation_df.reindex(columns=noised_df.columns)

# merging the two dataframes
merged_df = pd.concat([noised_df, validation_df], ignore_index=True)

# parse timestamps
merged_df["time:timestamp"] = pd.to_datetime(
    merged_df["time:timestamp"],
    errors="coerce",
    utc=True
)
```

```python
# convert back to event log
event_log = pm4py.convert_to_event_log(
    merged_df,
    case_id_key="case:concept:name",
    activity_key="concept:name",
    timestamp_key="time:timestamp"
)

# export event log as XES
pm4py.write_xes(event_log, "../data/recovered.xes")
```
