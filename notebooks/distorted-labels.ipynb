```python
import pandas as pd
import pm4py

log = pm4py.read_xes("../data/BPI Challenge 2017.xes.gz")
df = pm4py.convert_to_dataframe(log)
```

## Distorted Labels
Problem:
- two or more values are very similar (both syntactically and semantically)
- Effect: Discovered process models will overfit the event log, and be too little generic (as with “Polluted Labels”)

Manifestation & Detection:
- Minor differences in the letters of some attribute values
- E.g., “Further Information Required” and “Further Information Req.”

```python
print(df["concept:name"].unique())
```

```python
import random
import re

def distort_label(x):
    r = random.random()

    # no change (important for realism)
    if r < 0.4:
        return x

    # lowercase distortion
    if r < 0.55:
        return x.lower()

    # remove first character
    if r < 0.7:
        return x[1:] if len(x) > 1 else x

    # replace separators
    if r < 1.0:
        return re.sub(r"[_\- ]", "", x)

    return x

df["concept:name"] = df["concept:name"].astype(str).apply(distort_label)
```
